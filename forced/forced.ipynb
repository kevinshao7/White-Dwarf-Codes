```python
import numpy as np 
Temparr = np.arange(10000,30001,2500)
densarr = np.arange(0.1,1.1,0.1)
print(Temparr)
print(densarr)

for i in range(1):
    for j in range(1):
        dens = densarr[i]
        Temp=Temparr[j]
        # Replace text in a file
        file_path = 'forced_base.in'  # Path to your file
        newfile = "forced_{}_{}.in".format(dens,Temp)
        old_temp = "variable T0 equal 15000"
        new_temp = "variable T0 equal {}".format(Temp)
        old_dens = "variable density equal 1.0 #g/cm^3"
        new_dens = "variable density equal {} #g/cm^3".format(dens)
        old_dump = "dump mydmp Si custom 1000 traj.txt id type vx vy vz"
        new_dump = "dump mydmp Si custom 1000 traj_{}_{}.txt id type vx vy vz".format(dens,Temp)
        old_log = "log forcedlog"
        new_log = "log forcedlog_{}_{}".format(dens,Temp)
        # Read the file
        with open(file_path, 'r', encoding='utf-8') as file:
            content = file.read()
        # Replace the text
        updated_content = content.replace(old_temp, new_temp)
        updated_content = updated_content.replace(old_dens, new_dens)
        updated_content = updated_content.replace(old_dump, new_dump)
        updated_content = updated_content.replace(old_log, new_log)
        # Write the modified content back to the file (overwrite)
        with open(newfile, 'w', encoding='utf-8') as file:
            file.write(updated_content)
        file_path = 'forced_base.sh'  # Path to your file
        newfile = "forced_{}_{}.sh".format(dens,Temp)
        oldlog = "#SBATCH -o forced_baselog.txt"
        newlog = "#SBATCH -o forced_{}_{}.txt".format(dens,Temp)
        oldcmd = "mpiexec -n 24 lmp_mpi -in forced_base.in"
        newcmd = "mpiexec -n 24 lmp_mpi -in forced_{}_{}.in".format(dens,Temp)
        with open(file_path, 'r', encoding='utf-8') as file:
            content = file.read()
        # Replace the text
        updated_content = content.replace(oldlog, newlog)
        updated_content = updated_content.replace(oldcmd, newcmd)
        # Write the modified content back to the file (overwrite)
        with open(newfile, 'w', encoding='utf-8') as file:
            file.write(updated_content)


        
```
