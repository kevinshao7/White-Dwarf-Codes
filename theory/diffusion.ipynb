```python
T =15000
NA = 6.022e23
mSi = 0.0280855/NA
mH = 1.008/NA # in grams
n = 1# in gcc
n = 1e6*n/(1000*mH) #in atoms/m^3
kb = 1.380649e-23
#Vanderwaals radius
rH = 110e-12
rSi = 210e-12
pi=3.1415926535897
sigma = pi*(rH+rSi)**2
D = ((kb*T/(pi*mSi))**0.5)*2/(3*n*sigma)
d=1000
print("diffusion timescale in years")
print((d**2/D)/(3600*24*365.25))
print("mean free path")
print(1/(n*sigma*(2**0.5)))
print("interatomic separation")
print(n**(-1/3))
```

```python
#Screening 
me = 9.11e-31 #electron mass
e0 = 8.854e-12
kb = 1.380649e-23
hbar = 1.05e-34
pi = 3.14159
NA = 6.022e23
T = 15000
n = 1.0*1e6/(1.008/NA) #1gcc
q = 1.6022e-19 #electron charge
k0 = ((4*me*(q**2)*(3*n*pi**2)**(1/3))/(pi*(hbar**2)))**(1/2)
print("k0 in m^-1")
print(k0)
print("k0 in cm^-1")
print(k0/100)
print("Thomas-Fermi screening length")
print(1/k0)
#1 Joule is 1e7 erg
print("Energy in erg")
print(1e7*(q**2)/(4*pi*e0))
```
