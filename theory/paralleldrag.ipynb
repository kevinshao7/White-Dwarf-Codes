```python
import numpy as np 
from pydos2unix import dos2unix
import scipy as scp
#use 30 cpus to maximize usage
#for f in *.sh; do sbatch "$f"; done
velarr = np.logspace(2,6,50)

for i in range(len(velarr)):
    vel = velarr[i]
        # Replace text in a file
    file_path = 'dragbase.py'  # Path to your file
    newfile = "drag_{:.1e}.py".format(vel)
    old_vel = "vb = 1e4"
    new_vel = "vb={:.1e}".format(vel)
# Read the file
    with open(file_path, 'r', encoding='utf-8') as file:
        content = file.read()
    # Replace the text
    updated_content = content.replace(old_vel, new_vel)

    # Write the modified content back to the file (overwrite)
    with open(newfile, 'w', encoding='utf-8') as file:
        file.write(updated_content)
    with open(newfile, "rb") as src:
        buffer = dos2unix(src)
    with open(newfile, "wb") as dest:
        dest.write(buffer)

    file_path = 'dragbase.sh'  # Path to your file
    newfile = "drag_{:.1e}.sh".format(vel)
    oldlog = "#SBATCH -o drag_baselog.txt"
    newlog = "#SBATCH -o drag_{:.1e}_log.txt".format(vel)
    oldcmd = "python dragbase.py"
    newcmd = "python drag_{:.1e}.py".format(vel)
    with open(file_path, 'r', encoding='utf-8') as file:
        content = file.read()
    # Replace the text
    updated_content = content.replace(oldlog, newlog)
    updated_content = updated_content.replace(oldcmd, newcmd)
    # Write the modified content back to the file (overwrite)
    with open(newfile, 'w', encoding='utf-8') as file:
        file.write(updated_content)
    with open(newfile, "rb") as src:
        buffer = dos2unix(src)
    with open(newfile, "wb") as dest:
        dest.write(buffer)
```
