```python
import matplotlib.pyplot as plt
import numpy as np
import os

folder = "2026_02_04"  # path to your folder
datalist=[]
vel = []
colors=["red","green","blue","orange"]
for filename in os.listdir(folder):
    path = os.path.join(folder, filename)
    vel.append(filename.split("_")[1].split["."])
    datalist.append(np.load(path))

for i in range(len(datalist)):
    for j in range(4):
        vel = fil
        plt.scatter(datalist[i][j],color=colors[j])
```
